```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Machine-time required for each product
# Rows = machines
# Columns = products

A = np.array([
    [2, 1],   # Machine 1: Product A needs 2 hours, Product B needs 1 hour
    [1, 2]    # Machine 2: Product A needs 1 hour, Product B needs 2 hours
], dtype=float)

# Available time on each machine
b = np.array([100, 80], dtype=float)

# Production quantities you want
x = np.array([30, 40], dtype=float)
```

```python
# Calculate the machine time used
used = A @ x

# Calculate ranks
rank_A = np.linalg.matrix_rank(A)
rank_augmented = np.linalg.matrix_rank(
    np.column_stack((A, b))
)

print("Coefficient matrix A:")
print(A)

print("\nAvailable time b:")
print(b)

print("\nProduction quantities x:")
print(x)

print("\nMachine time used:")
print(used)

print("\nRank of A =", rank_A)
print("Rank of augmented matrix [A|b] =", rank_augmented)

if rank_A == rank_augmented:
    print("\nThe system is CONSISTENT.")
else:
    print("\nThe system is INCONSISTENT.")
```

```python
machines = [f"Machine {i+1}" for i in range(len(b))]

plt.figure(figsize=(7, 4))

plt.bar(machines, b, label="Available time")
plt.bar(machines, used, label="Used time")

plt.ylabel("Hours")
plt.title("Factory Machine-Time Usage")
plt.legend()

plt.show()
```
